```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
stock = yf.Ticker("AAPL")

data = stock.history(period="2y")

data.head()
```

```python
data["Daily_Return"] = data["Close"].pct_change()

data["Price_Change"] = data["Close"] - data["Open"]

data["Price_Change_Pct"] = (
    data["Close"] - data["Open"]
) / data["Open"]

data["High_Low_Range"] = (
    data["High"] - data["Low"]
) / data["Close"]

data[[
    "Open",
    "High",
    "Low",
    "Close",
    "Daily_Return",
    "Price_Change",
    "Price_Change_Pct",
    "High_Low_Range"
]].tail()
```

```python
data["MA_5"] = data["Close"].rolling(window=5).mean()
data["MA_20"] = data["Close"].rolling(window=20).mean()
data["MA_50"] = data["Close"].rolling(window=50).mean()

data["Price_to_MA20"] = data["Close"] / data["MA_20"]
data["Price_to_MA50"] = data["Close"] / data["MA_50"]

data[[
    "Close",
    "MA_5",
    "MA_20",
    "MA_50",
    "Price_to_MA20",
    "Price_to_MA50"
]].tail()
```

```python
data["Return_5D"] = data["Close"].pct_change(periods=5)

data["Return_20D"] = data["Close"].pct_change(periods=20)

data["Momentum_10D"] = (
    data["Close"] - data["Close"].shift(10)
)

data[[
    "Close",
    "Return_5D",
    "Return_20D",
    "Momentum_10D"
]].tail()
```

```python
data["Volatility_5D"] = data["Daily_Return"].rolling(window=5).std()

data["Volatility_20D"] = data["Daily_Return"].rolling(window=20).std()

data[[
    "Daily_Return",
    "Volatility_5D",
    "Volatility_20D"
]].tail()
```

```python
data["Volume_MA_20"] = data["Volume"].rolling(window=20).mean()

data["Volume_Ratio"] = (
    data["Volume"] / data["Volume_MA_20"]
)

data[[
    "Volume",
    "Volume_MA_20",
    "Volume_Ratio"
]].tail()
```

```python
data["High_Volume_Flag"] = (
    data["Volume_Ratio"] > 1.5
).astype(int)

data[[
    "Volume_Ratio",
    "High_Volume_Flag"
]].tail()
```

```python
delta = data["Close"].diff()

gain = delta.clip(lower=0)
loss = -delta.clip(upper=0)

average_gain = gain.rolling(window=14).mean()
average_loss = loss.rolling(window=14).mean()

relative_strength = average_gain / average_loss

data["RSI_14"] = (
    100 - (100 / (1 + relative_strength))
)

data[["Close", "RSI_14"]].tail()
```

```python
ema_12 = data["Close"].ewm(
    span=12,
    adjust=False
).mean()

ema_26 = data["Close"].ewm(
    span=26,
    adjust=False
).mean()

data["MACD"] = ema_12 - ema_26

data["MACD_Signal"] = data["MACD"].ewm(
    span=9,
    adjust=False
).mean()

data["MACD_Histogram"] = (
    data["MACD"] - data["MACD_Signal"]
)

data[[
    "Close",
    "MACD",
    "MACD_Signal",
    "MACD_Histogram"
]].tail()

```

```python
data["Price_Range"] = (
    data["High"] - data["Low"]
) / data["Close"]

data[[
    "High",
    "Low",
    "Close",
    "Price_Range"
]].tail()
```

```python
data["Open_Close_Return"] = (
    data["Close"] - data["Open"]
) / data["Open"]

data[[
    "Open",
    "Close",
    "Open_Close_Return"
]].tail()
```

```python
data["High_Low_Pct"] = (
    data["High"] - data["Low"]
) / data["Low"]

data[[
    "High",
    "Low",
    "High_Low_Pct"
]].tail()
```

```python
feature_columns = [
    "Close",
    "Daily_Return",
    "MA_5",
    "MA_20",
    "MA_50",
    "Price_to_MA20",
    "Price_to_MA50",
    "Return_5D",
    "Return_20D",
    "Momentum_10D",
    "Volatility_5D",
    "Volatility_20D",
    "Volume_Ratio",
    "High_Volume_Flag",
    "RSI_14",
    "MACD",
    "MACD_Signal",
    "MACD_Histogram",
    "Price_Range",
    "Open_Close_Return",
    "High_Low_Pct"
]

data[feature_columns].tail()
```

```python
missing_values = data.isnull().sum()

missing_values = missing_values[
    missing_values > 0
].sort_values(ascending=False)

missing_values
```

```python
data_clean = data.dropna().copy()

print("Original shape:", data.shape)
print("Cleaned shape:", data_clean.shape)
print("Remaining missing values:", data_clean.isnull().sum().sum())
```

```python
data_clean["Target_Return_1D"] = (
    data_clean["Close"].shift(-1) / data_clean["Close"]
) - 1

data_clean[[
    "Close",
    "Target_Return_1D"
]].tail()
```

```python
data_clean["Target_Up"] = (
    data_clean["Target_Return_1D"] > 0
).astype(int)

data_clean[[
    "Close",
    "Target_Return_1D",
    "Target_Up"
]].tail()
```

```python
data_clean = data_clean.dropna(
    subset=["Target_Return_1D"]
).copy()

print("Final dataset shape:", data_clean.shape)
print("Missing target values:")
print(data_clean[[
    "Target_Return_1D",
    "Target_Up"
]].isnull().sum())
```

```python
print(data_clean["Target_Up"].value_counts())

print("\nTarget proportions:")
print(
    data_clean["Target_Up"]
    .value_counts(normalize=True)
    .sort_index()
)
```

```python
target_columns = [
    "Target_Return_1D",
    "Target_Up"
]

X = data_clean[feature_columns].copy()
y = data_clean["Target_Up"].copy()

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
print("Feature columns:")
print(X.columns.tolist())
```

```python
split_index = int(len(data_clean) * 0.8)

X_train = X.iloc[:split_index].copy()
X_test = X.iloc[split_index:].copy()

y_train = y.iloc[:split_index].copy()
y_test = y.iloc[split_index:].copy()

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)

print("\nTraining period:")
print(data_clean.index[0], "to", data_clean.index[split_index - 1])

print("\nTesting period:")
print(data_clean.index[split_index], "to", data_clean.index[-1])
```

```python
print("Training target distribution:")
print(y_train.value_counts(normalize=True).sort_index())

print("\nTesting target distribution:")
print(y_test.value_counts(normalize=True).sort_index())
```

```python
from sklearn.metrics import accuracy_score

majority_class = y_train.mode()[0]

baseline_predictions = [
    majority_class
] * len(y_test)

baseline_accuracy = accuracy_score(
    y_test,
    baseline_predictions
)

print("Majority class:", majority_class)
print(f"Baseline accuracy: {baseline_accuracy:.2%}")
```
